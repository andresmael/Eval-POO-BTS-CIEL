## Partie 2 : Conception et implémentation de la classe `CMoteur` 

On souhaite désormais enrichir le projet en modélisant le moteur du véhicule via une nouvelle classe nommée **`CMoteur`**, définie dans les fichiers **`moteur.h`** et **`moteur.cpp`**.

### Cahier des charges de la classe `CMoteur`

#### 1. Attributs privés (`private`)
* `cylindree` (`float`) : cylindrée du moteur en litres (ex: `1.6`, `2.0`).
* `puissance` (`int`) : puissance fiscale ou DIN en chevaux (ex: `110`).
* `typeCarburant` (`std::string`) : type d'énergie (`"Essence"`, `"Diesel"`, `"Electrique"`, `"Hybride"`).
* `enMarche` (`bool`) : état du moteur (`true` = démarré, `false` = arrêté).
* `regime` (`int`) : régime moteur en tours/minute (`tr/min`).

#### 2. Méthodes publiques (`public`)
* **Constructeur par défaut `CMoteur()`** : initialise la cylindrée à `1.2`, la puissance à `75`, le carburant à `"Essence"`, `enMarche` à `false` et le `regime` à `0`.
* **Constructeur paramétré `CMoteur(float cyl, int puiss, std::string carb)`** : initialise les attributs correspondants, met `enMarche` à `false` et `regime` à `0`.
* **Destructeur `~CMoteur()`** : affiche un message de destruction (utile pour le suivi mémoire lors des tests).
* **`void demarrer()`** : si le moteur est arrêté, passe `enMarche` à `true` et fixe le `regime` au ralenti à `900` tr/min. S'il est déjà démarré, affiche un message d'avertissement.
* **`void arreter()`** : passe `enMarche` à `false` et remet le `regime` à `0` tr/min.
* **`void accelerer(int trMin)`** : augmente le `regime` de `trMin` **uniquement si le moteur est démarré** (sans dépasser un régime maximal de `6500` tr/min).
* **`int getPuissance()`** : accesseur (getter) retournant la puissance du moteur.
* **`void afficher()`** : affiche dans la console toutes les caractéristiques du moteur ainsi que son état actuel (Démarré/Arrêté et régime).

### Travail demandé
1. Créez le fichier d'en-tête **`moteur.h`** dans votre répertoire de projet VS Code (avec les directives anti-double inclusion `#ifndef MOTEUR_H`).
2. Créez le fichier d'implémentation **`moteur.cpp`** (n'oubliez pas d'inclure `"moteur.h"`).
3. Copiez votre code ci-dessous pour archivage dans le notebook.

````c++
// Collez ici le contenu de votre fichier moteur.h
#ifndef MOTEUR_H
#define MOTEUR_H

#include <string>

class CMoteur
{
private:
    float cylindree;
    int puissance;
    std::string typeCarburant;
    bool enMarche;
    int regime;

public:
    CMoteur(1.2, 75, "Essence", false, 0);
    CMoteur(float cyl, int puiss, std::string carb);
    ~CMoteur();

    void demmarer();
    void arreter();
    void accelerer(int trMin);
    int getPuissance();
    void afficher();
};

#endif
````

````c++
// Collez ici le contenu de votre fichier moteur.cpp
#include "moteur.h"
#include <iostream>

CMoteur::CMoteur(float _cylindree, int _puissance, std::string _typeCarburant, bool _enMarche, int _regime) {
    cylindree = _cylindree;
    typeCarburant = _typeCarburant;
    enMarche = _enMarche;
    regime = _regime;
}

CMoteur::~CMoteur() {}

void CMoteur::demmarer() {
    if (enMarche = false) {
        enMarche = true;
        regime = 900;
        std::cout << "Le moteur a bien démmarer";
    }
    else {
        std::cout << "Erreur le moteur est déja démmarer";
    }
}

void CMoteur::arreter() {
    if (enMarche = true) {
        enMarche = false;
        regime = 0;
        std::cout << "Le moteur c'est bien arréter";
    }
    else {
        std::cout << "Erreur le moteur est déja arréter";
    }
}

void CMoteur::accelerer(int trMin) {
    if (enMarche = true) {
        if ((regime + trMin) <= 6500) {
            regime += trMin;
        }
        else {
            regime = 6500;
        }
    }
    else {
    std::cout << "Erreur le moteur est arréter";
    }
}

void CMoteur::getPuissance() {
    if (enMarche = true) {
    std::cout << "Le régime du moteur est de"<< regime << "tr/min";
    }
    else {
    std::cout << "Le régime est de 0 tr/min";
    }
}

void CMoteur::afficher() {
    std::cout << "Le cylindre est de :" << cylindree
              << "La puissance est de :" << puissance
              << "Le type de carburant est :" << typeCarburant
              << "Le régime du moteur est de :" << regime;
    if (enMarche = true) {
        std::cout << "Le moteur est en marche"
    }          
    else {
        std::cout << "Le moteur est en éteint"
    }    
}
````
